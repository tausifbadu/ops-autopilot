```pyspark
transformer_monthly_profile = spark.sql ("""SELECT
    transformer_id,
    date_trunc('month', day_ts) AS month_ts,

    ROUND(SUM(morning), 2)   AS morning_sum,
    ROUND(SUM(afternoon), 2) AS afternoon_sum,
    ROUND(SUM(evening), 2)   AS evening_sum,
    ROUND(SUM(night), 2)     AS night_sum,

    ROUND(SUM(total), 2)     AS month_total

FROM `electric_analysis_dev`.`transformer_parts_of_day_analysis`

GROUP BY
    transformer_id,
    date_trunc('month', day_ts)

ORDER BY
    transformer_id,
    month_ts""")

spark.sql("create database if not exists `electric_analysis_dev`")


transformer_monthly_profile.coalesce(1).write \
    .mode("overwrite") \
    .format("parquet") \
    .option("path", "s3://ops-autopilot-data/transformed/transformer_monthly_profile/") \
    .saveAsTable("`electric_analysis_dev`.`transformer_monthly_profile`")
```
