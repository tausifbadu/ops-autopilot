```pyspark
sc.install_pypi_package("matplotlib")
sc.install_pypi_package("pandas")
sc.install_pypi_package("boto3")
```

```pyspark
df = spark.table("`electric_analysis_dev`.`avg_hourly_load`")
```

```pyspark
pdf = df.toPandas()

import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))
plt.plot(pdf["hour"], pdf["avg_usage"], marker="o")
plt.xticks(range(0,24))
plt.title("Average Hourly Load")
plt.grid(True)

plt.savefig("hourly_load.png")
print("Chart saved")
```

```pyspark
import boto3

s3 = boto3.client("s3")

s3.upload_file(
    "hourly_load.png",
    "ops-autopilot-data",
    "charts/avg_hourly_load.png"
)

print("Uploaded to S3")
```
